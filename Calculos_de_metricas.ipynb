```python
import numpy as np
from scipy.io.wavfile import read
from scipy.stats import pearsonr

print("| Mensaje secreto | SNR (dB) | Correlación de Pearson |")
print("|---|---:|---:|")

for i in range(1, 6):
    _, x = read(f'secret{i}.wav')
    _, x_hat = read(f'recovered_{i}.wav')

    x = x.astype(np.float64)
    x_hat = x_hat.astype(np.float64)

    if x.ndim > 1:
        x = x[:, 0]
    if x_hat.ndim > 1:
        x_hat = x_hat[:, 0]

    # 1. SNR (dB)
    potencia_senal = np.sum(x ** 2)
    potencia_error = np.sum((x - x_hat) ** 2)
    snr = 10 * np.log10(potencia_senal / potencia_error)

    # 2. Correlación de Pearson
    corr, _ = pearsonr(x, x_hat)

    print(f"| Secret {i} | {snr:.4f} | {corr:.6f} |")
```
